The `open` function opens a file - it provides an object through which you can access the file's contents.

````python
f = open('data/hamlet.txt')  # open a file for reading
````

````python
f  # What's this?
````

````python
lines = f.readlines()  # Read the entire file as a list of lines
lines[:7]  # Note, that each string ends with the newline character (\n)
````

````python
f.close()  # it's important to close the file!
````

There are often situations when we need to create an object for a moment, use it, and then close/delete/release it.  Most often, it is a resource (file, network connection, synchronization object, etc.), the longer retention of which or incorrect release of which could have negative consequences.

This is so common that Python has a special syntax structure to easily deal with this situation: the `with` statement (More information: https://docs.python.org/2/reference/compound_stmts.html#the-with-statement).

This is the recommended way of operating on files (when opening and closing them "manually", various libraries usually have data loading functions that do not require opening the files manually). We want to manually open the file and load the contents, so we can do it like this:

````python
with open('data/hamlet.txt') as f:  # f contains the file object, as before
    for _ in range(10):
        line = f.readline()  # Read a line
        line_cleaned = line.strip()  # Removes leading and trailing whitespaces (incl. the newline character)
        print(line_cleaned)

# when leaving the with block (either normally or by exception), f will be closed automatically
````

````python
# Alternative, shorter version:
with open('data/hamlet.txt') as f:  # f contains the file object, as before
    for i, line in enumerate(f):  # Iterating over file object returns the contents line by line
        print(line.strip())
        if i >= 10:
            break  # we only want first 10 lines
````

Remember about line breaks - e.g. `readline` will load a line with a line ending character ending it (`this is a line\n`). Line breaks and other whitespace can be removed using the `strip()` method of string objects.

````python
'to jest linia\n'
````

````python
'to jest linia\n'.strip()
````

Line breaks (`'\n'`) must be added manually (see documentation, e.g. `write`, `writelines`).

Worth reading: https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files

## File access modes

The most common file access modes:
* 'r' - Read access (default). File is read from the beginning to the end.
* 'w' - Write access. File is created **or truncated if it alredy exists**, and writen from scratch.
* 'a' - Append mode = write access, but new contents is added to the end of the file. 

````python
# Let's first create some data:
from word_tools import word_counts

with open('./data/hamlet.txt') as f:
    hamlet_file_contents = f.read()  # read the entier file in one go:
    hamlet_word_counts = word_counts(hamlet_file_contents)

hamlet_word_counts
````

````python
with open('./outputs/hamlet_word_counts.txt', 'w') as f:
    for word, count in hamlet_word_counts:
        f.write(f'{word} : {count}\n')  # writing to a file (Note: you need to manually add a line break - \n)
````

**Note** - The physical writing does not happen when the `write` function is used, but from time to time - when python decides that it has collected enough data to write. In particular, it is possible that the save will take place only when closing the file.

I recommend the documentation: https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files

## Browsing/traversing files and directories

If we want to list files in one directory:

````python
import os  # os  = (o)perating (s)ystem utilities

os.listdir('./data/L202_example_1')
````

A more advanced method is the `walk` function from the `os` module (https://docs.python.org/3/library/os.html#os.walk), analyze an example of its use:

````python
import os

for root, dirs, files in os.walk('./data/L202_example_1'):
    print('#' * 40)
    print('root:', root)    
    print('dirs:', dirs)
    print('files:', files)
````
