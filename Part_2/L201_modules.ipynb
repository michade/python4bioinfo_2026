# Modules, packages, python beyond anaconda

(hamlet.py demonstration)

Python code is organized into **modules** and **packages**.

Modules are just files with the `.py` extension. They can be either _executed_ as scripts using python (more or less like a regular program), or _imported_ by other python or Jupyter files.
Importing allows to use the functions and variables defined in the module.

A larger unit of organization, packages, are (in broad overview) directories containing modules, recognized by Python by the presence of a file called `__init__.py` in the directory (it contains the package initialization code, it may be empty).

```python
import word_tools  # I am importing a module

word_tools.word_counts('To be, or not to be?')
```

```python
import word_tools as wt # I am importing a module under an alias

wt.word_counts('To be, or not to be?')
```

```python
from word_tools import word_counts  # I am importing a specific module element

word_counts('To be, or not to be?')
```

```python
from word_tools import *  # I import all objects (need to be carefull)

print(word_counts('To be, or not to be?'))
print(make_words_list('To be, or not to be?', '?,'))
```

## Where does Python get its modules from?

In general, Python first searches the current directory, then the user's local Python directory, and finally the system directory (if the two are distinguished).

More information: https://docs.python.org/3/tutorial/modules.html#the-module-search-path

```python
import sys

sys.path  # Python search path
```

The "\__pycache\__" directory contains are auxiliary, precompiled code files that Python creates to speed up work when loading modules. They can be safely removed.
