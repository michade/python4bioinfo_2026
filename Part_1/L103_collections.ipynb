# Introduction to Python - Collections

**Collections** are objects that store other objects (more precisely: references to other objects).

Sequential types store elements in a specific order. Each element has its own _index_, starting from 0 (this is due to the convention in older languages like C, where the index meant the offset from the beginning of the collection)

The basic sequence types are tuples and lists. Strings can also be treated as sequences of single characters.

## 1. Tuples

A tuple is a way to store a small, fixed, number of other python objects together.

For example a point in 3D space:

````python
point = (-1.0, 4.1, 2.4)
point
````

Or some arbitrary data:

````python
my_tuple = (34, 2.1, "Cat in a hat", 7)
my_tuple
````

some basic operations:

````python
len(my_tuple)  # number of elements in the tuple
````

To access elements of _any collection_ we use square brackets:

````python
my_tuple[1]
````

Note, that **the first element of the collection has index 0**:

````python
my_tuple[0]
````

Negative values of the index mean **counting from the end, with -1 being the last element**:

````python
my_tuple[-1]
````

We can also **unpack** values of a tuple (or other collection for that matter):

````python
print(point)

x, y, z = point    # "unpacking" a tuple
print('x ==', x)
print('y ==', y)
print('z ==', z)
````

### 2. Lists

The _list_ is the most common collection type in Python, made to store an arbitrary number of other objects.
Lists, in contrast to tuples or strings, can be changed after creation (more on this later).

````python
my_list = [145, 2.0, "Dog in the fog", 3, 5]
````

Lists support the same operations as tuples:

````python
print( my_list[2] )
print( my_list[-2] )
print( len(my_list) )
````

**"slices"** are used to select a fragment of a list (or other collection): `list[start:end:step]`

* `start` - index of the first element
* `end` - index limiting from above
* `step` - optional, means taking "some" element

````python
x = [10, 20, 30, 40, 50, 60, 70, 80, 90]
````

````python
x[2:5]  # try changing the start and stop.
````

Notice that e.g. for slice 2:5 the element with `end` index (5) **is not included**

Some more examples:

````python
print(x[:4])  # you can skip the index -> it will be replaced by 0...
print(x[2:])  # or the length of the list
````

````python
print(x[1:-3])  # you can use negative indices
print(x[-4:-1])
````

````python
x[5:2]  # empty!
````

The `step` parameter allows to e.g. take every n-th element, or go in reverse. Notice how start and stop work!

````python
x[1:9:2]
````

````python
x[9:1:-1]
````

An easy way to create a list containing a sequence of numbers is to use the `range` function. The geral form is:

```python
range(start, end, step)
```

Arguments are like in _slices_:
* `start` - first number
* `end` - the first number that will **not** be included.
* `step` - optional, means taking "some" element. If ommited will be set to 1.

````python
range(2, 7)
````

For now it will be easier to understand, if we convert it to a list:

````python
list(range(2, 7))
````

````python
# Works a bit like slices:
print(list( range(5) ))
print(list( range(2, 7) ))
print(list( range(3, 15, 2) ))
print(list( range(10, 5, -1) ))
````

To find element in a list we can use `in` operatror, or the `index` _attribute_ (recall the basic operations):

````python
my_list
````

````python
print( 3 in my_list )
print( "Dog in the fog" in my_list )
print( 123 in my_list )
````

````python
print( my_list.index(3) )
print( my_list.index("Dog in the fog") )
# print( my_list.index(123) )  # This line would generate an error
````

**But note the following! :**

````python
"3" in my_list
````

### Modyfying lists

Unlike tuples, we can add elements to an already existing list:

````python
print(my_list)
my_list.append("The end.")
my_list.insert(2, "I am now at index 2")
print(my_list)
````

...as well as remove them, either by index, or by value:

````python
my_list.remove("Dog in the fog")  # Remove by value
print(my_list)
````

````python
last_item = my_list.pop()  # Remove the last item. It will also be returned from the pop() call.
print(my_list)
print(last_item)
````

````python
value = my_list.pop(2)  # Remove the item from index 2. It will also be returned from the pop() call.
print(my_list)
print(value)
````

### Interlude: Mutable and immutable types

Types in Python are divided into _mutable_ and _immutable_ :

* _immutable_ - objects of this type cannot be changed. Examples: `int`, `str`, `tuple`
* _mutable_ - an object can be changed/updated. Examples: `list`, `dict`

Consider and example: suppose I have some items in my office at the MiNI department, and I want to put them on a list for inventory purposes.

````python
stuff_in_my_office = ["Chair", "Whiteboard", "Desk", "Kettle"]   # (1)
````

I share this office with my co-worker, Sebastian, who also needs a list like that.
Since it is the same room:

````python
stuff_in_sebastians_office = stuff_in_my_office   # (2)
````

Now imagine my faculty generously sponsored a coffe machine for my team. I will put it in my office:

````python
stuff_in_my_office.append("Coffe machine")
print(stuff_in_my_office)
````

Does Sebastian have a Coffe machine in his room now?

````python
print(stuff_in_sebastians_office)
````

This is because `stuff_in_my_office` and `stuff_in_sebastians_office` are **physically the same list**, that we created in line (1).
We then, in line (2), simply created another label, that refers to the same list.

To be technically precise: we operate on objects in Python using _references_: `stuff_in_my_office` is a reference to some object (list) in computer memory,
which stores all these items. In line (2) we do **not** create a copy of this object, we just _copy the reference_ to this object.

Contrast with this:

````python
list_a = ["Chair", "Whiteboard", "Desk", "Kettle"] 
list_b = list_a.copy()  # Notice the .copy() here!!!
list_a.append("Coffe machine")
print(list_a)
print(list_b)
````

### Object identity

  * The `is` operator tests the _identity_ of objects: whether they are physically the same object
  * The `==` operator tests the _equivalence_ of objects (whether the )

Consider the following examples:

````python
print(stuff_in_my_office)
print(stuff_in_sebastians_office)

print(stuff_in_my_office == stuff_in_sebastians_office)  # We have the same stuff in our rooms...
print(stuff_in_my_office is stuff_in_sebastians_office)  # ...because we, in fact, have the same room
````

````python
list_a = ["Chair", "Whiteboard", "Desk", "Kettle"] 
list_b = list_a.copy()
print(list_a)
print(list_b)

print(list_a == list_b)  # The two lists' contents are identical...
print(list_a is list_b)  # ...but they are separate lists and can be changed separately.
````

What about `tuple` - which is an immutable type and cannot be changed after creation (no append, pop, etc.)?

````python
tuple_a = (2, "carrot", 3.5)
tuple_b = tuple(tuple_a)  # a way to copy a tuple
tuple_c = (2, "carrot", 3.5)

# of course:
print(tuple_a == tuple_b)  # (4)
print(tuple_a == tuple_c)  # (5)
print(tuple_b == tuple_c)  # (6)

# ...but what about the follwing?
print(tuple_a is tuple_b)
print(tuple_a is tuple_c)
print(tuple_b is tuple_c)
````

Now, handling identity (`is`) in case of immutable types is a bit complicated. However, it is usually (except for some very rare cases), irrelevant.

**Final, important, note on tuples** - while the contents of a tuple cannot change, the tuple can contain e.g. a list, **which can be modified**. This is because the tuple stores not the physical list itself **but the reference to the list**. The same, in fact, goes for all collections.

````python
my_tuple = (2, "carrot", [1, 2, 3])
identical_tuple = (2, "carrot", [1, 2, 3])
copy_of_my_tuple = tuple(my_tuple)
print(my_tuple)
print(copy_of_my_tuple)
print(identical_tuple)
````

Let's add something to the list stored in my tuple:

````python
my_tuple[2].append(-7)
print(my_tuple)
print(copy_of_my_tuple)
print(identical_tuple)
````
