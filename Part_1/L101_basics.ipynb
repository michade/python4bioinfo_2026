## Writing Python code

### What is code?
So... what is "code" in a "programming language" (Python) anyway, and how does it translate to apps running on my computer? 
How can I solve my data-procesing problems by writing and running code?

The **code** ("source code") of a program ("program" is the word old people like me use to call apps) is a detailed descrition
of the work that the computer is supposed to perform, and how it is supposed to behave.

The **programming language** (here: Python) is a particular syntax, that this description conforms to.
In Python this can mostly be thourght as a sequence of operations (this is called _imperative programming_).
The source code is, in itself, just text. In order for the code to be able to **do** something you needs additional tools:
If you have Python installed in your system, it can take your python code file (or a jupyter notebook) and run it, i.e.
read the sequence of operations you described, and execute them as specified.

_Some more context: Every program you have ever used has some source code written in some language.
For the apps on your phone or computer usually a different language was used (like `C`, `Java` or `C#`).
You would not have the access to their source code (this belongs to the companies that made these programs),
instead it would be pre-processed by tools called **compilers** to a form that can be easily distributed for your particualr system - like `.exe` files in Windows, or binary executable files in Linux and MacOS.
In modern consumer systems (laptops, PCs, Macs, phones), this is further automated - you just "install the program from the store", which means the system just downloads these binary files for you, and sets them up.
In summary, in "everyday life" you don't deal with source code or binary files, but they always exist. In general sense, the source code of a program  **is** the program.

We can write _expressions_ that combine operations on objects:



````python
2 + 3
````

````python
7 / 2  # Division
````

Expressions can also include functions. For example, `abs` is a built-in function that returns the absolute value of a number (more theory on function calls later, for now just accept this intuitively):

````python
2 + abs(-2)  # abs() is a built-in function that returns the absolute value of a number
````

A function call is an expression that consists of a function name followed by parentheses containing zero or more arguments.

In _imperative programming languages_ (like Python) a function is not a mapping (like in mathemathics - another false friend),
but a request for the computer to perform some computation, and _return_ (produce) a value, i.e. some object.

When this expression is evaluated, the abs function is called with the object `-2` as its argument, and it returns the object `2` (an `int`).

````python
abs(-2)
````

````python
type(abs(-2))
````

````python
max(3, 2, 7)
````

Besides arithmethic operators (`+`, `-`, `*`, `/`, `//` etc.), we also have an important class of comparison operators: `==`, `!=`, `>`, `>=`, `<`, etc.

They might seem different, but actually work the same they take their operands and produce an object of type `bool` (logical), that represents the results:

````python
print(2 * 3 + 1 == 7)  # Note the double equal sign == is used for comparison, while a single equal sign = is used for assignment (see below)
print(2 != 3)  # Inequality operator
print(2 > 3)  # Greater than
print(2 + 2 <= 4)  # Less than or equal to
````

To combine logical operations use the operators `and` and `or`:

````python
3 > 2 and 1 == 0
````

````python
False or 2 + 2 == 4
````

### Variables

Objects can be _assigned_ (using `=`) to variables, so that we can refer to them and manipulate them later:

````python
x = 7  # Assign the value 7 to the variable x
print(x)  # Print the value of x (strictly speaking: the object, to which x refers)

y = (x + 3) * abs(x - 10)
print(y)
````

We can later refer to the value of x and y in other cells:

````python
print(x)
print(y)

# Note, we can alway redefine x (assign other object to this label)
x = "Haha! I'm a string now!"
print(x)
````

Let's make sure we understand assignment:

````python
x = 6
y = 3
x = y + 1
print(x)
````

**Important remark:**

The line `x = y + 1` **NOT** a statement of methematical fact or an equation!

It is a **command** telling Python to:
1. Compute `y + 1`, which produces the object `4`
2. Assign this object to the variable `x`

**Important reminder:**

Remeber the difference between `=` and `==`!!!

## More strings and string interpolation

There are 2 alternative ways to specify string literals: using `"` and `'`. They are absolutely equivalent.

````python
print("That's one way to make a string")
print('And this is another, but now I can use "quotes" inside.')
````

To easily create strings containing the values of expressions or variables, you can use "string interpolation":

1. Add the letter `f` **before** the first apostrophe, like so: `f"`, or `f'`.
2. Then, in the string, you can put an expression in brackets - it's value will be calculated an inserted in this location.

See these examples:

````python
print(f"This is the expression: {2 + 2} we want to print")
x = 320
print(f'This is the expression: {x / 10} we want to print')
print('Forgot the `f`! {2 + 2}')
````

Strings can include special characters, like tabulation, or the _newline_ characters, which cause the text to be broken into lines when using `print`.
To specify them, we use the backslash character `\`; The letter that follows it will determine it's special meaning.

````python
print("Some \t tabulation \t characters.")
````

````python
print("A newline causes the text...\n..to appear in a new line. Like this:\nEnd.")
````

Use `\\` to print the backslash (`\`) itself.

````python
print("\\ \\\\")
````

## The "assignment by _(operation)_" statements

There is one last type of expression, that beginners need to learn: the "assignment by _(operation)_".

For example:
```python
x += 3
```
...is equivalent to:
```python
x = x + 3
```

The same goes for most other arithmethic operators:

````python
x = 2
print('x =', x)
x += 5
print('x =', x)
x *= 10
print('x =', x)
x -= 13
print('x =', x)
````
