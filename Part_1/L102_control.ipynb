# Introduction to Python - Control flow

### The `if` statement

The `if` statement (or the `if-else` statement) provides the ability to conditionally execute code based on the values of some expressions.

````python
x = 9 # Try different values for x
y = 0
if x + 2 > 10:   # condition
    print("greater than 10")   # code block (indented - 4 spaces/tab)
    y = 20
print("this statement always executes")
print(y)
````

````python
x = 3  # Try different values for x
if x > 3:
    print("greater than 3")   # code block (indented - 4 spaces/tab)
else:
    print("NOT greater than 3")
````

Sidenote: To get an empty block we use an empty "pass" statement

````python
if x > 3:
    pass
````

To test several conditions sequentially, we use `elif` (short for "else if")

_(Python does not have the "switch" statement known from some languages)_

````python
x = "one"

if x == "zero":
    y = 0
elif x == "one":  # "else, if..."
    y = 1
elif x == "two":
    y = 2
elif x == "one":  # duplicated
    y = 1000
elif x == "three":
    y = 3
else:
    y = None
    print("greater than 3")
    
print(y)
````

There is also a 3-nary operator `if-else`, which evaluates to one of two provided values, based on the value of an expression.

````python
x = "Yes" if 2 < 3 else "No"
x
````

````python
abs(-3) if x == 'Yes' else (3 + 2) * 5
````

### The `for` statement ("for loop")

The `for` statement in Python is the most common type of a _loop_ - a statement that allows to consecutively execute the same piece of code
with changing variable values.

Initially we will cover some basic examples, using the `range` function. More advanced use cases will appear when we will talk about _collection_ types.

````python
for number in range(6):
    print("A number: ", number)  # Note, that the end (6) is excluded - it's a Python convention
````

````python
# There can be (and often is) more than one expression in the loop block:
for x in range(3): 
    print('x =', x)
    y = x + 1
    print('...and y =', y)

print()
print("In Python, x will be availible after the loop as well:")
print("x = ", x)
````

Note, that the following loop:

````python
z = 0
for i in range(3):
    z = z + 1  # This line is executed multiple times
print("z =", z)
````

...would be equivalent to the code below:

````python
z = 0
z = z + 1
z = z + 1
z = z + 1
print("z =", z)
````

More options to work with ranges:

````python
print('Providing just the end value:')
for a in range(3):
    print(a)

print('Providing start and end values:')
for a in range(4, 8):
    print(a)

print('Providing start, end, and step values:')
for a in range(15, 45, 10):
    print(a)

print('Decreasiong order (note the last value):')
for a in range(3, 0, -1):
    print(a)

print('Positive step and end <= start results in an empty range:')
for a in range(3, 3):
    print(a)
print('(Nothing got printed above)')
````

The keywords `break` and `continue` are used to modify the execution of the loop.

* The `break` statement terminates the loop execution.
* The `continue` statement skips the rest of the lines _in the current interation_.

````python
for i in range(20):
    print(f"Starting execution for i == {i}")
    print("  Do some stuff...")
    if i == 3:
        break
    print("  Do some more stuff...")
    print(f'Done for i == {i}')
print("Loop finished")
````

````python
for i in range(4):
    print(f"Starting execution for i == {i}")
    print("  Do some stuff...")
    if i == 2:
        continue
    print("  Do some more stuff...")
    print(f'Done for i == {i}')
print("Loop finished")
````

````python
# This won't work, only numbers of type int are allowed!
# for i in range(2.5, 6.5):
#     print(i)

# We have to work around:
for i in range(2, 6):
    print(i + 0.5)
````

**Important note 1:** in the Python's for loop, the loop variable (here: `i`) is reset at the beginning of each iteration:

````python
for i in range(4):  # In this line `i` gets the next value from the range, regardless of its current value
    print('Iteration start: i = ', i)
    i = 8
    print('Iteration end: i = ', i)
````

### The `while` statement

The `while` loop is a very basic and elastic type of a loop.

It starts each iteration by checking is the provided condition is `True` - if so it will execute the indented code block;
Otherwise it finishes.

````python
a = 2  # We NEED to declare i before - the while loop does not 
b = 1
while a + b <= 7:
    print(f'Iteration start: a=={a}, b=={b}, a+b=={a+b}')
    a = a + 2
    print(f'Iteration end: a=={a}, b=={b}, a+b=={a+b}')
print(f"Loop finished, a=={a}, b=={b}")
````

````python
# `break` and `continue` can be used here as well:
i = 0
while i < 100:
    print('Iteartion start, i = ', i)
    i = i + 1
    print("  Do some stuff...")
    if i == 2:
        continue
    print("  Do some more stuff...")
    if i > 4:
        break
    print('Iteartion end, i = ', i)
    
print('Loop finished, i = ', i)
````

A common pattern is to make an _infinite_ `while` loop and use `break` to exit manually.
It's useful in case of complex loop logic, when we might want to check for the exit condition at a specific time.

````python
i = 0
while True:  # always passes
    i += 1
    print('i =', i)
    if i == 4:  # gotta put some break inside, or else it will spin forever!
        break
````

## Nesting control flow statements

Control flow statements can be nested. It it a completely usual practice, though too deep (let's say more than 3-4 levels) is discouraged because of decreasd readability.

````python
# Experiment with the values of the variables:
run_the_for_loop = True
x = 3

if run_the_for_loop:
    print("I'll run the for loop")
    for i in range(3):
        print(i)
else:
    print("Let's check if x is even...")
    if x % 2 == 0:
        print("Yep")
    else:
        print("Nope")
    print("Let's run some other loop:")
    while x > 0:
        x -= 1
        print("x =", x)
````
