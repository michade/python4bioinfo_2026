# Introduction to Python - defining functions

In Python `functions` are encolsed pieces of code, that perform some computation and return some value.
They can also have side effects - like printing something on the screen, or creating a plot.
We have already used some pre-defined _built-in_ functions (like `abs` or `max`).

We will now define our own functions, so we can use them in multiple places later.
This way our code is better organized.

The template for defining a custom function is as follows:
```python
def the_function_name(some_parameter, another_parameter, maybe_yet_another_parameter):
    # There can be 0, 1, 2 or more paramters
    # The body of the function goes here
    # To return a value we use:    
    return some_value
```

For example:

````python
def my_example_function(a, b):
    print(f'The function was called with a={a} and b={b}')
    print('These commands will be executed every time the funciton is called.')
    if a > 0:
        c = a + b * 2
    else:
        c = -1

# executing this cell will create the function, but nothing else will happen!
````

Now, let's _call_ the function (this will execute the commands inside the function):

````python
my_example_function(3, 10)
````

..and again:

````python
my_example_function(-3, 20)
````

````python
for x in range(-2, 3):
    print(my_example_function(x, 100))
````

## Side effects of functions

Functions in python have _side effects_.

````python
def add_1_to_numbers_in_list(lst):
    print('Printing this text is a side effect of running the function.')
    for i in range(len(lst)):
        lst[i] += 1
    return None  # to emphasize that nothing is returned here
    
my_list = [1, 2, 3]
print(my_list)
    
add_1_to_numbers_in_list(my_list)
print(my_list)

add_1_to_numbers_in_list(my_list)
print(my_list)
````

## Advanced function parameter specification

**Terminological node:**

Consider the function definition, and some function call:
```python
def fun(x):
    return x + 5

y = fun(123)
```

A _formal parameter_ (or just _parameter_) is the name which we use inside the function to refer to the passed value. In the example above it is `x`.

An _argument_ (or _actual parameter_) is the value (object) you pass to the function. i.e. here it is the object `123` (an `int`).

Right now there is much confusion in the materials online, many use these terms interchangeably.


### Positional and keyword parameters

Python functions can have "positional" (required) and "keyword" parameters/arguments.

Positional parameters are the ones we have seen so far.
They arguments (objects) passed are assigned to paramters in their order of passing (hence: "positional").
All must be specified when calling the function.

````python
def my_fun(a, b, c, d):
    return f'a:{a}, b:{b}, c:{c}, d:{d}'

my_fun(3, 5, 2, 6)
````

Keyword parameters can be added, by specifying their default value.
They are optional when calling the function, and can be specified in any order, afetr the positional ones, with the syntax: `parameter_name=value`.

````python
def my_fun(a, b, c=-1, d=10):
    return f'a:{a}, b:{b}, c:{c}, d:{d}'

my_fun(1, 2)
````

Optional parameters can be omitted...

````python
my_fun(1, 2, d=15)
````

Or passed according to position:

````python
my_fun(1, 2, 15)
````

*(Not recommended)* Positional parameters can also be passed by name

````python
my_fun(b=5, a=3)
````

## Functions with a variable number of arguments

A function which can have a varied number of arguments is called _variadic_.

````python
def my_fun_2(arg1, arg2, *args):
    print(args)
    
my_fun_2(12, 34, 56, 78)
````

Beware of mixing optional arguments and the `*args` expression!

````python
def my_fun_3(arg1, arg2, kw1='ala', kw2='ma', *args):
    print(args)
    print(kw1)

my_fun_3(12, 34, kw1='foo')  # it works for now
````

````python
my_fun_3(12, 34, 56, kw2='foo')  # here 56 does not enter 'args' but as kw1
````

We can also enable a function to have any number of unspecified keyword parameters. They are accessible as a `dict`:

````python
def my_fun_4(arg1, kw1=-1, kw2=-2, **kwargs):
    print(arg1)
    print(kw1)
    print(kw2)
    print(kwargs)
    
my_fun_4(123, kw1=45, kw2=67, kw3=89, kw4='sometext')
````

### Universal function signature:

The following function signature (specification of parameters) can capture all possible ways to pass arguments:

````python
def my_universal_fun(*args, **kwargs):
    print(args)
    print(kwargs)
````

````python
my_universal_fun()
````

````python
my_universal_fun(1, 'aaa', my_arg=123, other_arg='foo')
````

````python
my_universal_fun(1, 'aaa', 456, other_arg='foo')
````

## Functions are objects

````python
def add_2(x):
    return x + 2

my_fun = add_2  # ...we can assign them to variables just like any other object
````

````python
my_fun
````

````python
my_fun(4)
````

````python
def apply_to_lst(lst, fun):  # we can pass functions as parameters to other functions
    return [fun(elem) for elem in lst]

apply_to_lst([10, 11, 12], my_fun)
````

## Anonymous functions (_lambda functions_, _lambda expressions_)

Anonymous functions are short functions that do not have a name. They are meant to be e.g. immediately passed to another function as parameters (though they can be assigned to variables as well).

````python
lambda x: x+2
````

````python
(lambda x: x+2)(7)
````

````python
apply_to_lst([3, 7, -1], lambda x: x ** 2)
````

````python
(lambda x, y: x + y)(5, 3)  # two-parameter lambda
````

Example use: _key_ parameter in functions `sorted` or `min`:

````python
my_list = [
    (2, 8), (1, 5), (2, 2), (1, -1), (3, 4)
]
````

````python
sorted(my_list)
````

````python
sorted(my_list, key=lambda x: x[1])
````

In the example below, notice:
1. `sorted` returns a sorted copy of the list, the `sort` method sorts (changes) the given object.
2. `key` is a function that assigns a sorting key to an object.
3. `lambda (x, y):...` has one parameter (2-element tuple). `lambda x, y:...` has two parameters.
4. `reverse` - means descending sorting.

````python
sorted(my_list, key=lambda (x, y): x + y, reverse=True)
````
